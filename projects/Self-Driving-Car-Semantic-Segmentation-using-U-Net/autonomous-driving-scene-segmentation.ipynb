# Reproducible training contract

Set `CITYSCAPES_DATA_ROOT` before running this notebook. The directory must contain non-empty `train/` and `val/` folders of image pairs. The notebook validates readability and filename overlap, never deletes source data, sets seeds, and writes generated artifacts under `ARTIFACT_DIR`.

This notebook has been reset to an unexecuted state. It contains no committed metrics; any results must come from a real dataset/model run.

```python
import os
import random
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

SEED = 42
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)

data_root_value = os.environ.get("CITYSCAPES_DATA_ROOT")
DATA_ROOT = Path(data_root_value).expanduser() if data_root_value else Path()
if not data_root_value or not DATA_ROOT.is_dir():
    raise FileNotFoundError(
        "Set CITYSCAPES_DATA_ROOT to an existing directory containing train/ and val/ "
        f"image-pair folders; received {data_root_value or '<unset>'}."
    )
TRAIN_PATH = DATA_ROOT / "train"
VAL_PATH = DATA_ROOT / "val"
for split_path in (TRAIN_PATH, VAL_PATH):
    if not split_path.is_dir():
        raise FileNotFoundError(f"Required dataset split is missing: {split_path}")

SUPPORTED_IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png"}
def image_files(folder):
    return sorted(
        p.name for p in folder.iterdir()
        if p.is_file() and p.suffix.lower() in SUPPORTED_IMAGE_EXTENSIONS
    )
train_files = image_files(TRAIN_PATH)
val_files = image_files(VAL_PATH)
if not train_files or not val_files:
    raise RuntimeError(f"Dataset splits must be non-empty: train={TRAIN_PATH}, val={VAL_PATH}")
overlap = set(train_files).intersection(val_files)
if overlap:
    raise RuntimeError(f"Data leakage risk: {len(overlap)} filenames occur in both train and val (examples: {sorted(overlap)[:5]})")

ARTIFACT_DIR = Path(os.environ.get("ARTIFACT_DIR", "artifacts")).expanduser()
ARTIFACT_DIR.mkdir(parents=True, exist_ok=True)
print(f"Training image pairs: {len(train_files)}")
print(f"Validation image pairs: {len(val_files)}")
print(f"Artifacts will be written to: {ARTIFACT_DIR.resolve()}")
```

```python
import tensorflow as tf
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.layers import Conv2D, Conv2DTranspose, Dropout, Input, MaxPooling2D, concatenate
from tensorflow.keras.models import Model

# TensorFlow seed is set after importing TensorFlow; deterministic kernels remain platform-dependent.
tf.random.set_seed(SEED)
print(f"Training images: {len(train_files)}")
print(f"Validation images: {len(val_files)}")
```

```python
def validate_images(folder_path, filenames):
    invalid = []
    for filename in filenames:
        image = cv2.imread(str(folder_path / filename), cv2.IMREAD_COLOR)
        if image is None or image.size == 0:
            invalid.append(f"{filename}: unreadable")
            continue
        if image.ndim != 3 or image.shape[2] != 3:
            invalid.append(f"{filename}: expected 3-channel image, got shape {image.shape}")
        if image.shape[1] < 2 or image.shape[1] % 2:
            invalid.append(f"{filename}: expected an even width >= 2 for image|mask pairs, got {image.shape[1]}")
    if invalid:
        raise RuntimeError(
            f"Invalid image pairs in {folder_path}: {invalid[:10]}"
            + (" ..." if len(invalid) > 10 else "")
        )
    print(f"Validated {len(filenames)} readable image pairs in {folder_path}")
validate_images(TRAIN_PATH, train_files)
validate_images(VAL_PATH, val_files)
```

```python
def check_dataset_balance(path, name, files):
    print(f"\nChecking {name} dataset")
    sizes = []
    for filename in files:
        image = cv2.imread(str(path / filename), cv2.IMREAD_COLOR)
        if image is not None:
            sizes.append(image.shape)
    print(f"Total images: {len(files)}")
    if not sizes:
        raise RuntimeError(f"No readable images found in {path}")
    unique_shapes = np.unique(np.asarray(sizes), axis=0)
    print(f"Unique image shapes: {len(unique_shapes)}")
    if len(unique_shapes) > 1:
        print("Dataset contains inconsistent image shapes:", [tuple(s) for s in unique_shapes])
    else:
        print("All images have consistent dimensions:", tuple(unique_shapes[0]))

check_dataset_balance(TRAIN_PATH, "Training", train_files)
check_dataset_balance(VAL_PATH, "Validation", val_files)
print("\nTrain/Validation ratio =", len(train_files), "/", len(val_files))
```

```python
def load_and_split_image(image_path):
    img = tf.io.read_file(image_path)
    img = tf.image.decode_image(img, channels=3, expand_animations=False)
    img = tf.cast(img, tf.float32)
    width = tf.shape(img)[1]
    tf.debugging.assert_greater_equal(width, 2, message="An image pair must be at least two pixels wide")
    tf.debugging.assert_equal(tf.math.floormod(width, 2), 0, message="An image pair must have an even total width")
    half_width = width // 2
    input_image = img[:, :half_width, :]
    mask_image = img[:, half_width:, :]
    input_image = input_image / 255.0
    mask_image = mask_image / 255.0
    return input_image, mask_image
print("The image loading and detachment function has been successfully built")
```

```python
rng = random.Random(SEED)
random_samples = rng.sample(train_files, min(5, len(train_files)))
for i, filename in enumerate(random_samples):
    full_path = TRAIN_PATH / filename
    input_img, mask_img = load_and_split_image(str(full_path))
    plt.figure(figsize=(14, 6))
    plt.subplot(1, 2, 1)
    plt.title(f"Sample {i + 1}: Original Image (Street)")
    plt.imshow(input_img)
    plt.axis("off")
    plt.subplot(1, 2, 2)
    plt.title(f"Sample {i + 1}: Segmentation Mask")
    plt.imshow(mask_img)
    plt.axis("off")
    plt.show()
```

```python
CLASS_NAMES = ["Void/Background", "Road", "Construction", "Object", "Nature", "Sky", "Human", "Vehicle"]

mapping_dict = {
    (128, 64, 128): 1, (244, 35, 232): 1, 
    (70, 70, 70): 2, (102, 102, 156): 2, (190, 153, 153): 2, 
    (153, 153, 153): 3, (250, 170, 30): 3, (220, 220, 0): 3,   
    (107, 142, 35): 4, (152, 251, 152): 4, 
    (70, 130, 180): 5,  
    (220, 20, 60): 6, (255, 0, 0): 6,     
    (0, 0, 142): 7, (0, 0, 70): 7, (0, 60, 100): 7, (0, 0, 230): 7, (119, 11, 32): 7 
}

def rgb_to_class(mask):
    """Map each mask pixel to its nearest supported Cityscapes grouping."""
    mask = tf.cast(tf.round(mask * 255), tf.int32)
    palette = tf.constant(list(mapping_dict), dtype=tf.int32)
    class_ids = tf.constant(list(mapping_dict.values()), dtype=tf.int32)
    distances = tf.reduce_mean(tf.abs(mask[..., tf.newaxis, :] - palette), axis=-1)
    nearest = tf.argmin(distances, axis=-1, output_type=tf.int32)
    minimum_distance = tf.reduce_min(distances, axis=-1)
    tf.debugging.assert_less(
        tf.reduce_max(minimum_distance), 40,
        message="Mask contains colors outside the supported palette/tolerance",
    )
    return tf.expand_dims(tf.gather(class_ids, nearest), axis=-1)
def process_data(image_path):
    img = tf.io.read_file(image_path)
    img = tf.image.decode_image(img, channels=3, expand_animations=False)
    img = tf.cast(img, tf.float32)
    width = tf.shape(img)[1]
    tf.debugging.assert_greater_equal(width, 2, message="An image pair must be at least two pixels wide")
    tf.debugging.assert_equal(tf.math.floormod(width, 2), 0, message="An image pair must have an even total width")
    w = width // 2
    input_image = img[:, :w, :]
    mask_image = img[:, w:, :]
    
    input_image = tf.image.resize(input_image, (256, 256))
    mask_image = tf.image.resize(mask_image, (256, 256), method='nearest')
    
    input_image = input_image / 255.0
    mask_image = rgb_to_class(mask_image / 255.0)
    
    input_image.set_shape([256, 256, 3])
    mask_image.set_shape([256, 256, 1])
    
    return input_image, mask_image

print("The functions have been corrected to be compatible with Graph Mode.")
```

```python
random_file = random.choice(train_files)
sample_path = os.path.join(TRAIN_PATH, random_file) 

inp, msk = process_data(sample_path)

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.title("Input Image")
plt.imshow(inp)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Processed Mask (Classes)")
plt.imshow(msk[:, :, 0] * 30, cmap='nipy_spectral') 
plt.axis('off')

plt.show()

unique_classes = np.unique(msk.numpy())
print(f" The numbers in this image (Classes Found): {unique_classes}")

if len(unique_classes) > 2:
    print("Test successful: The model successfully converted colors into multiple categories.")
else:
    print("Warning: The model did not find enough classes. Try running the code again on a different image.")
```

```python
BATCH_SIZE = 16
AUTOTUNE = tf.data.AUTOTUNE

def create_dataset(image_paths, is_training=True):
    dataset = tf.data.Dataset.from_tensor_slices([str(path) for path in image_paths])
    dataset = dataset.map(process_data, num_parallel_calls=AUTOTUNE)
    if is_training:
        dataset = dataset.shuffle(buffer_size=min(1000, len(image_paths)), seed=SEED, reshuffle_each_iteration=False)
    options = tf.data.Options()
    options.experimental_deterministic = True
    dataset = dataset.with_options(options).batch(BATCH_SIZE).prefetch(AUTOTUNE)
    return dataset

train_img_paths = [TRAIN_PATH / filename for filename in train_files]
val_img_paths = [VAL_PATH / filename for filename in val_files]
train_dataset = create_dataset(train_img_paths, is_training=True)
val_dataset = create_dataset(val_img_paths, is_training=False)
print("Deterministic dataset pipelines built.")
for x, y in train_dataset.take(1):
    print(f"Dataset ready: images {x.shape}, masks {y.shape}")
```

```python
for images, masks in train_dataset.take(1):
    print("\n Checking the dimensions of the batch(Batch Shape Check):")
    print(f"   - Images Shape: {images.shape} ")
    print(f"   - Masks Shape:  {masks.shape} ")
    
    if images.shape[-1] == 3 and masks.shape[-1] == 1:
        print("\n The dimensions are perfectly accurate. We are ready to build the model. (U-Net)!")
    else:
        print("\n There is an error in the dimensions; please refer to the previous steps..")
```

```python
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, Dropout, Conv2DTranspose, concatenate
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau

def build_unet_model(input_shape=(256, 256, 3), num_classes=8):
    inputs = Input(input_shape)

    c1 = Conv2D(64, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(inputs)
    c1 = Dropout(0.1)(c1)
    c1 = Conv2D(64, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(c1)
    p1 = MaxPooling2D((2, 2))(c1)

    c2 = Conv2D(128, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(p1)
    c2 = Dropout(0.1)(c2)
    c2 = Conv2D(128, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(c2)
    p2 = MaxPooling2D((2, 2))(c2)

    c3 = Conv2D(256, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(p2)
    c3 = Dropout(0.2)(c3)
    c3 = Conv2D(256, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(c3)
    p3 = MaxPooling2D((2, 2))(c3)

    c4 = Conv2D(512, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(p3)
    c4 = Dropout(0.2)(c4)
    c4 = Conv2D(512, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(c4)
    p4 = MaxPooling2D(pool_size=(2, 2))(c4)

    c5 = Conv2D(1024, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(p4)
    c5 = Dropout(0.3)(c5)
    c5 = Conv2D(1024, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(c5)

    u6 = Conv2DTranspose(512, (2, 2), strides=(2, 2), padding='same')(c5)
    u6 = concatenate([u6, c4])
    c6 = Conv2D(512, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(u6)
    c6 = Dropout(0.2)(c6)
    c6 = Conv2D(512, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(c6)

    u7 = Conv2DTranspose(256, (2, 2), strides=(2, 2), padding='same')(c6)
    u7 = concatenate([u7, c3])
    c7 = Conv2D(256, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(u7)
    c7 = Dropout(0.2)(c7)
    c7 = Conv2D(256, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(c7)

    u8 = Conv2DTranspose(128, (2, 2), strides=(2, 2), padding='same')(c7)
    u8 = concatenate([u8, c2])
    c8 = Conv2D(128, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(u8)
    c8 = Dropout(0.1)(c8)
    c8 = Conv2D(128, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(c8)

    u9 = Conv2DTranspose(64, (2, 2), strides=(2, 2), padding='same')(c8)
    u9 = concatenate([u9, c1])
    c9 = Conv2D(64, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(u9)
    c9 = Dropout(0.1)(c9)
    c9 = Conv2D(64, (3, 3), activation='relu', kernel_initializer='he_normal', padding='same')(c9)

    outputs = Conv2D(num_classes, (1, 1), activation='softmax')(c9)

    model = Model(inputs=[inputs], outputs=[outputs])
    return model

model = build_unet_model()

model.compile(optimizer='adam', 
              loss='sparse_categorical_crossentropy', 
              metrics=['accuracy'])

print(" The model was successfully built and assembled.")
model.summary()
```

```python
checkpoint_path = ARTIFACT_DIR / "cityscapes_unet_model.keras"
callbacks = [
    ModelCheckpoint(str(checkpoint_path), verbose=1, save_best_only=True, monitor="val_loss"),
    EarlyStopping(patience=5, monitor="val_loss", restore_best_weights=True),
    ReduceLROnPlateau(monitor="val_loss", factor=0.1, patience=3, verbose=1),
]
print("Training is starting; no metrics are reported until this cell completes.")
history = model.fit(train_dataset, validation_data=val_dataset, epochs=150, callbacks=callbacks)
print(f"Training complete. Best model artifact: {checkpoint_path.resolve()}")
```

```python
def get_metrics(dataset_name, dataset, model, num_classes=8):
    print(f"Calculating measurements for {dataset_name}...")
    loss, accuracy = model.evaluate(dataset, verbose=0)
    confusion = np.zeros((num_classes, num_classes), dtype=np.int64)
    for images, masks in dataset:
        predictions = tf.argmax(model(images, training=False), axis=-1).numpy().reshape(-1)
        targets = tf.cast(masks, tf.int32).numpy().reshape(-1)
        valid = (targets >= 0) & (targets < num_classes)
        confusion += np.bincount(
            targets[valid] * num_classes + predictions[valid],
            minlength=num_classes * num_classes,
        ).reshape(num_classes, num_classes)
    true_positive = np.diag(confusion).astype(np.float64)
    union = confusion.sum(axis=1) + confusion.sum(axis=0) - true_positive
    dice_denominator = confusion.sum(axis=1) + confusion.sum(axis=0)
    present = union > 0
    iou = np.divide(true_positive, union, out=np.zeros(num_classes), where=present)
    dice = np.divide(true_positive * 2, dice_denominator, out=np.zeros(num_classes), where=dice_denominator > 0)
    return {
        "loss": float(loss),
        "pixel_accuracy": float(accuracy),
        "mean_iou": float(iou[present].mean()) if np.any(present) else 0.0,
        "mean_dice": float(dice[dice_denominator > 0].mean()) if np.any(dice_denominator > 0) else 0.0,
        "per_class_iou": dict(zip(CLASS_NAMES, iou.tolist())) if "CLASS_NAMES" in globals() else iou.tolist(),
        "per_class_dice": dict(zip(CLASS_NAMES, dice.tolist())) if "CLASS_NAMES" in globals() else dice.tolist(),
        "confusion_matrix": confusion,
    }

train_metrics = get_metrics("Training Data", train_dataset, model)
val_metrics = get_metrics("Validation Data", val_dataset, model)
summary = pd.DataFrame(
    {
        "Metric": ["Pixel Accuracy", "Loss", "Mean IoU (present classes)", "Mean Dice (present classes)"],
        "Training": [train_metrics["pixel_accuracy"], train_metrics["loss"], train_metrics["mean_iou"], train_metrics["mean_dice"]],
        "Validation": [val_metrics["pixel_accuracy"], val_metrics["loss"], val_metrics["mean_iou"], val_metrics["mean_dice"]],
    }
)
print(summary.to_string(index=False))
```

```python
def create_colored_mask(mask):
    colors = np.array([
        [0, 0, 0], [128, 64, 128], [70, 70, 70], [250, 170, 30],
        [107, 142, 35], [70, 130, 180], [220, 20, 60], [0, 0, 142]
    ])
    mask_indices = tf.squeeze(mask, axis=-1).numpy()
    mask_indices = np.clip(mask_indices, 0, 7)
    return colors[mask_indices].astype(np.uint8)

def show_predictions(dataset, num=3):
    plt.figure(figsize=(15, 5 * num))
    for image, mask in dataset.take(1):
        pred_mask = model.predict(image, verbose=0)
        pred_mask = tf.argmax(pred_mask, axis=-1)
        pred_mask = tf.expand_dims(pred_mask, axis=-1)
        available = int(image.shape[0])
        for i in range(min(num, available)):
            plt.subplot(num, 3, i*3 + 1)
            plt.title("Original Image")
            plt.imshow(image[i])
            plt.axis('off')
            
            plt.subplot(num, 3, i*3 + 2)
            plt.title("True Mask")
            plt.imshow(create_colored_mask(mask[i]))
            plt.axis('off')
            
            plt.subplot(num, 3, i*3 + 3)
            plt.title("AI Prediction")
            plt.imshow(create_colored_mask(pred_mask[i]))
            plt.axis('off')
    plt.show()

print(" Visualizing Predictions...")
show_predictions(val_dataset, num=3)
```

```python
print("Generating demo video...")
output_video_path = ARTIFACT_DIR / "cityscapes_demo.avi"
fourcc = cv2.VideoWriter_fourcc(*"MJPG")
frame_size = (256, 256)
video_writer = cv2.VideoWriter(str(output_video_path), fourcc, 10, frame_size)
if not video_writer.isOpened():
    raise RuntimeError(f"Could not open video artifact for writing: {output_video_path}")
colors = np.array([
    [0, 0, 0], [128, 64, 128], [70, 70, 70], [250, 170, 30],
    [107, 142, 35], [70, 130, 180], [220, 20, 60], [0, 0, 142],
], dtype=np.uint8)
try:
    for count, (image, _) in enumerate(val_dataset.unbatch().take(100), start=1):
        pred_mask = tf.argmax(model(tf.expand_dims(image, axis=0), training=False), axis=-1)[0].numpy()
        colored_mask_bgr = cv2.cvtColor(colors[pred_mask], cv2.COLOR_RGB2BGR)
        original_img_bgr = cv2.cvtColor((image.numpy() * 255).astype(np.uint8), cv2.COLOR_RGB2BGR)
        overlay = cv2.addWeighted(original_img_bgr, 0.6, colored_mask_bgr, 0.4, 0)
        video_writer.write(cv2.resize(overlay, frame_size))
        if count % 20 == 0:
            print(f"Processed {count} frames")
finally:
    video_writer.release()
print(f"Video saved: {output_video_path.resolve()}")
```
